```python
# INVENTORY MANAGEMENT SYSTEM

# Step 1: Initial stock from user
Rice = int(input("Enter rice stock: "))
Sugar = int(input("Enter sugar stock: "))
Oil = int(input("Enter oil stock: "))

# Step 2: Loop for multiple operations (using for loop)
for i in range(10):   # allows up to 10 operations
    print("\n**** WELCOME TO THE INVENTORY MANAGEMENT SYSTEM ****")
    print("1. Select Product")
    print("2. Exit")
    
    choice = int(input("Enter your option: "))
    
    

    if choice == 1:   # Product selection
        print("Product List")
        print("1. Rice")
        print("2. Sugar")
        print("3. Oil")

        
        product = int(input("Enter product number: "))

        if product == 1:
            product_name = "Rice"
            Stock = Rice
        elif product == 2:
            product_name = "Sugar"
            Stock = Sugar
        elif product == 3:
            product_name = "Oil"
            Stock = Oil
        else:
            print("Invalid Product")
            continue

        # Step 3: Stock operations
        print("**** Stock Operation ****")
        print("1. Add Stock")
        print("2. Sell Stock")
        print("3. Check Stock")

        operation = int(input("Enter the operation: "))

        if operation == 1:   # Add stock
            qty = int(input("Enter quantity to add: "))
            Stock = Stock + qty
            if product == 1:
                Rice = Stock
            elif product == 2:
                Sugar = Stock
            else:
                Oil = Stock
            print("Stock added successfully")
            print("Current Stock:", Stock)

        elif operation == 2:   # Sell stock
            qty = int(input("Enter quantity to sell: "))
            if qty <= Stock:
                Stock = Stock - qty
                if product == 1:
                    Rice = Stock
                elif product == 2:
                    Sugar = Stock
                else:
                    Oil = Stock
                print("Stock sold successfully")
                print("Current Stock:", Stock)
            else:
                print("Not enough stock available")
                print("Available Stock:", Stock)

        elif operation == 3:   # Check stock
            print("Product:", product_name)
            print("Current Stock:", Stock)

        else:
            print("Invalid Operation entered")

    elif choice == 2:   # Exit
        print("Program Ended")
        break

    else:
        print("Invalid option")
```
